```python
horas = int(input())
if horas <= 2:
    valor = 8.0
elif horas <= 5:
    valor = 8.0 + (horas - 2) * 3.0
else:
    valor = 17.0 + (horas - 5) * 2.0
print(f"R$ {valor:.2f}")
```

```python
total = 0.0
while total < 25.0:
    valor = int(input())
    total += valor
troco = total - 25.0
print(f"Total inserido: R$ {total:.2f}")
print(f"Troco: R$ {troco:.2f}")
```

```python
maior = None
posicao = -1
for i in range(1, 9):
    num = int(input())
    if maior is None or num > maior:
        maior = num
        posicao = i
print(f"Maior: {maior}")
print(f"Posição: {posicao}")
```

```python
maior = None
segundo_maior = None
for _ in range(8):
    num = int(input())
    if maior is None or num > maior:
        segundo_maior = maior
        maior = num
    elif segundo_maior is None or num > segundo_maior:
        segundo_maior = num
print(f"Maior: {maior}")
print(f"Segundo maior: {segundo_maior}")
```

```python
anterior = None
repetido = False
for _ in range(8):
    num = int(input())
    if anterior is not None and num == anterior:
        repetido = True
    anterior = num
if repetido:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

```python
saldo = 500.0
for _ in range(6):
    tipo = int(input())
    valor = float(input())
    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")
print(f"Saldo final: R$ {saldo:.2f}")
```

```python
valor = int(input())
cedulas = [100, 50, 20, 10, 5, 2, 1]
for c in cedulas:
    qtd = valor // c
    valor %= c
    print(f"{qtd} cédula(s) de R$ {c}")
```

```python
num = int(input())
e_primo = True
if num <= 1:
    e_primo = False
else:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            e_primo = False
            break
if e_primo:
    print("É primo")
else:
    print("Não é primo")
```

```python
n = int(input())
qtd = 0
for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        qtd += 1
print(f"Quantidade de divisores: {qtd}")
```

```python
a = int(input())
b = int(input())
qtd_primos = 0
for num in range(a, b + 1):
    if num > 1:
        e_primo = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                e_primo = False
                break
        if e_primo:
            print(num)
            qtd_primos += 1
print(f"Total de primos encontrados: {qtd_primos}")
```

```python
secreto = 37
tentativas = 0
acertou = False
while tentativas < 5 and not acertou:
    palpite = int(input())
    tentativas += 1
    if palpite == secreto:
        print("Você acertou!")
        acertou = True
    elif palpite < secreto:
        print("O número secreto é maior.")
    else:
        print("O número secreto é menor.")
```

```python
saldo = 200.0
entradas = 0
retiradas_realizadas = 0
retiradas_recusadas = 0
total_entrou = 0.0
total_saiu = 0.0

while True:
    codigo = int(input())
    if codigo == 0:
        break
    valor = float(input())
    if codigo == 1:
        saldo += valor
        entradas += 1
        total_entrou += valor
    elif codigo == 2:
        if valor <= saldo:
            saldo -= valor
            retiradas_realizadas += 1
            total_saiu += valor
        else:
            retiradas_recusadas += 1

print(f"Saldo final: R$ {saldo:.2f}")
print(f"Quantidade de entradas: {entradas}")
print(f"Quantidade de retiradas realizadas: {retiradas_realizadas}")
print(f"Quantidade de retiradas recusadas: {retiradas_recusadas}")
print(f"Total que entrou: R$ {total_entrou:.2f}")
print(f"Total que saiu: R$ {total_saiu:.2f}")
```
