```python
numeros = []
for _ in range(10):
    numeros.append(int(input()))

pares = []
impares = []

for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)

print(numeros)
print(pares)
print(impares)
```

```python
numeros = []
for _ in range(8):
    numeros.append(int(input()))

maior = numeros[0]
menor = numeros[0]
pos_maior = 0
pos_menor = 0

for i in range(1, len(numeros)):
    if numeros[i] > maior:
        maior = numeros[i]
        pos_maior = i
    if numeros[i] < menor:
        menor = numeros[i]
        pos_menor = i

print(maior, pos_maior)
print(menor, pos_menor)
```

```python
nomes = []
while True:
    nome = input()
    if nome == "FIM":
        break
    if nome not in nomes:
        nomes.append(nome)

print(nomes)
print(len(nomes))
```

```python
dias = ("Domingo", "Segunda-feira", "Terça-feira", "Quarta-feira", "Quinta-feira", "Sexta-feira", "Sábado")
opcao = int(input())

if 1 <= opcao <= 7:
    print(dias[opcao - 1])
else:
    print("Opção inválida")
```

```python
lista_temp = []
for _ in range(6):
    lista_temp.append(int(input()))

numeros = tuple(lista_temp)

qtd_cinco = 0
pos_cinco = -1
qtd_pares = 0

for i in range(len(numeros)):
    if numeros[i] == 5:
        qtd_cinco += 1
        if pos_cinco == -1:
            pos_cinco = i
    if numeros[i] % 2 == 0:
        qtd_pares += 1

print(qtd_cinco)
if pos_cinco != -1:
    print(pos_cinco)
else:
    print("Não encontrado")
print(qtd_pares)
```

```python
aluno = {}
aluno['nome'] = input()
aluno['turma'] = input()
aluno['nota1'] = float(input())
aluno['nota2'] = float(input())

aluno['media'] = (aluno['nota1'] + aluno['nota2']) / 2

if aluno['media'] >= 7:
    aluno['situacao'] = "Aprovado"
elif aluno['media'] >= 5:
    aluno['situacao'] = "Recuperação"
else:
    aluno['situacao'] = "Reprovado"

print(aluno)
```

```python
frequencia = {}
for _ in range(8):
    palavra = input()
    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1

for palavra, qtd in frequencia.items():
    print(f"{palavra}: {qtd}")
```

```python
estoque = {
    "Caneta": 100,
    "Caderno": 50,
    "Borracha": 30,
    "Lápis": 80
}

prod = input()
qtd = int(input())

if prod not in estoque:
    print("Produto não cadastrado")
else:
    if estoque[prod] >= qtd:
        estoque[prod] -= qtd
        print("Venda realizada com sucesso")
    else:
        print("Estoque insuficiente")

print(estoque)
```

```python
turma = []
for _ in range(5):
    aluno = {}
    aluno['nome'] = input()
    n1 = float(input())
    n2 = float(input())
    aluno['notas'] = [n1, n2]
    aluno['media'] = (n1 + n2) / 2
    turma.append(aluno)

for aluno in turma:
    print(aluno['nome'], aluno['media'])
```

```python
catalogo = [
    (101, "Camiseta", 49.90),
    (102, "Calça", 89.90),
    (103, "Tênis", 199.90),
    (104, "Boné", 29.90),
    (105, "Meia", 9.90)
]

codigo_busca = int(input())
encontrado = False

for prod in catalogo:
    if prod[0] == codigo_busca:
        print(prod[1], prod[2])
        encontrado = True
        break

if not encontrado:
    print("Código não encontrado")
```

```python
estoque = [
    {"codigo": 1, "nome": "Teclado", "quantidade": 10, "preco": 150.0},
    {"codigo": 2, "nome": "Mouse", "quantidade": 20, "preco": 80.0},
    {"codigo": 3, "nome": "Monitor", "quantidade": 5, "preco": 900.0},
    {"codigo": 4, "nome": "Headset", "quantidade": 15, "preco": 200.0},
    {"codigo": 5, "nome": "Webcam", "quantidade": 8, "preco": 250.0}
]

print("Todos os itens:")
for item in estoque:
    print(item)

valor_total = 0
for item in estoque:
    valor_total += item["quantidade"] * item["preco"]

print("Valor total do estoque:", valor_total)

maior_valor_item = estoque[0]
for item in estoque:
    valor_item = item["quantidade"] * item["preco"]
    valor_maior = maior_valor_item["quantidade"] * maior_valor_item["preco"]
    if valor_item > valor_maior:
        maior_valor_item = item

print("Item com maior valor em estoque:", maior_valor_item)

resumo = (len(estoque), valor_total)
print("Resumo:", resumo)
```
