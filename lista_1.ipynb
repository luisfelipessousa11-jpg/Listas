```python
pares = 0
for _ in range(10):
    num = int(input())
    if num % 2 == 0:
        pares += 1
print(f"Foram digitados {pares} números pares.")
```

```python
secreto = 7
tentativas = 0
while True:
    chute = int(input())
    tentativas += 1
    if chute == secreto:
        print("Parabéns! Você acertou.")
        break
    elif chute < secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")
print(f"Tentativas: {tentativas}")
```

```python
qtd = 0
positivos = 0
negativos = 0
soma = 0

while True:
    num = int(input())
    if num == 0:
        break
    qtd += 1
    soma += num
    if num > 0:
        positivos += 1
    elif num < 0:
        negativos += 1

print(f"Quantidade: {qtd}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Soma: {soma}")
```
